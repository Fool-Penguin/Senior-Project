# Filter repositories with all guidance files

This notebook keeps only repositories that have `SECURITY.md`, `CONTRIBUTING.md`, `SKILL.md`, and `AGENTS.md` somewhere in their Git tree.

```python
from pathlib import Path

import pandas as pd

INPUT_CSV = Path("repo_metadata.csv")
OUTPUT_CSV = Path("repos_with_all_guidance.csv")
REQUIRED_COLUMNS = [
    "Has_SECURITY",
    "Has_SKILL",
    "Has_AGENTS",
    "Has_CONTRIBUTING",
]
```

```python
repos = pd.read_csv(INPUT_CSV)

missing_columns = set(REQUIRED_COLUMNS) - set(repos.columns)
if missing_columns:
    raise ValueError(f"Missing required columns: {sorted(missing_columns)}")

# CSV values are text, so normalize them before filtering.
has_all_guidance_files = repos[REQUIRED_COLUMNS].apply(
    lambda column: column.astype(str).str.strip().str.lower().eq("true")
)
filtered_repos = repos.loc[has_all_guidance_files.all(axis=1)].copy()

print(f"Selected {len(filtered_repos):,} of {len(repos):,} repositories.")
filtered_repos
```

```python
filtered_repos.to_csv(OUTPUT_CSV, index=False, encoding="utf-8-sig")
print(f"Saved {len(filtered_repos):,} repositories to {OUTPUT_CSV}")
```
